```python
# Membaca input dari pengguna (atau menggunakan nilai default)
jumlah_buku = int(input("Masukkan jumlah buku: "))
harga_per_buku = float(input("Masukkan harga per buku: "))
persen_diskon = float(input("Masukkan persentase diskon (%): "))
uang_budi = float(input("Masukkan uang Budi: "))
jumlah_orang = int(input("Masukkan jumlah orang untuk patungan: "))

# 1. Perhitungan Total Belanja, Diskon, dan Jumlah yang Harus Dibayar
total_belanja = jumlah_buku * harga_per_buku
besar_diskon = total_belanja * (persen_diskon / 100)
jumlah_bayar = total_belanja - besar_diskon

# 2. Uang Kembalian
kembalian = uang_budi - jumlah_bayar

# 3. Patungan 4 orang secara adil (bilangan bulat dan sisa rupiah)
per_orang = int(jumlah_bayar // jumlah_orang)
sisa_rupiah = int(jumlah_bayar % jumlah_orang)

# 4. Apakah uang Budi cukup? (True / False)
uang_cukup = uang_budi >= jumlah_bayar

# Menampilkan Output
print(f"Total       : {int(total_belanja)}")
print(f"Diskon      : {int(besar_diskon)}")
print(f"Bayar       : {int(jumlah_bayar)}")
print(f"Kembalian   : {int(kembalian)}")
print(f"Per orang   : {per_orang}")
print(f"Sisa        : {sisa_rupiah}")
print(f"Uang cukup? : {uang_cukup}")
```

```python
# Membaca input dari pengguna
jumlah_panel = int(input("Masukkan jumlah panel surya: "))
daya_panel_wp = float(input("Masukkan daya per panel (Wp): "))
jam_penyinaran = float(input("Masukkan rata-rata penyinaran efektif (jam/hari): "))
efisiensi = float(input("Masukkan efisiensi sistem (misal 0.8): "))
kebutuhan_listrik = float(input("Masukkan kebutuhan listrik rumah (kWh/hari): "))
tarif_listrik = float(input("Masukkan tarif listrik per kWh: "))

# 1. Energi yang dihasilkan per hari (kWh)
energi_harian = (jumlah_panel * daya_panel_wp / 1000) * jam_penyinaran * efisiensi

# 2. Energi per bulan (30 hari) dan nilai rupiah listrik yang dihasilkan
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif_listrik

# 3. Kekurangan energi per hari terhadap kebutuhan
kekurangan_energi = kebutuhan_listrik - energi_harian

# 4. Apakah PLTS sudah mencukupi kebutuhan rumah? (True / False)
mencukupi = energi_harian >= kebutuhan_listrik

# Menampilkan Output dengan fungsi round() untuk merapikan angka desimal
print(f"Energi harian (kWh) : {round(energi_harian, 1)}")
print(f"Energi bulanan (kWh): {round(energi_bulanan, 1)}")
print(f"Nilai listrik (Rp)  : {round(nilai_listrik, 1)}")
print(f"Kekurangan (kWh)    : {round(kekurangan_energi, 1)}")
print(f"Mencukupi?          : {mencukupi}")
```
