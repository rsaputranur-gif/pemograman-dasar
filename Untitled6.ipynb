```python
# 1. Contoh penggunaan %d (Integer / Bilangan Bulat)
jumlah_porsi = 3
print("Hendi memesan %d porsi mi ayam." % jumlah_porsi)

# 2. Contoh penggunaan %f (Float / Bilangan Pecahan/Desimal)
harga = 15000.50
print("Harga satu porsi adalah Rp %f" % harga)
# (Opsional: untuk membatasi 2 angka di belakang koma bisa menggunakan "%.2f" % harga)

# 3. Contoh penggunaan %x (Hexadecimal / Basis 16)
nomor_meja = 255
print("Nomor meja dalam hexadecimal: %x" % nomor_meja)  # Menghasilkan 'ff'

# 4. Contoh penggunaan %o (Octal / Basis 8)
jumlah_kursi = 10
print("Jumlah kursi dalam octal: %o" % jumlah_kursi)  # Menghasilkan '12'
```

```python
# 1. Mengubah total_detik agar menerima inputan dari keyboard
# Menggunakan int() karena input() mengembalikan string, sedangkan kita butuh angka untuk operasi matematika
total_detik = int(input("Masukkan total detik: "))

# Atau jika langsung menguji nilai 358200 sesuai nomor 2:
# total_detik = 358200

# 2. Proses konversi waktu
jam = total_detik // 3600
sisa_detik = total_detik % 3600
menit = sisa_detik // 60
detik = sisa_detik % 60

# 3. Menampilkan hasil
print(f"Hasil Konversi: {jam} Jam, {menit} Menit, {detik} Detik")
```

```python
# 1 & 2. Menerima input dari keyboard untuk standar kelulusan dan 5 komponen nilai mahasiswa
standar_kelulusan = float(input("Masukkan standar kelulusan: "))

print("Masukkan 5 komponen nilai mahasiswa:")
nilai1 = float(input("Nilai Tugas 1: "))
nilai2 = float(input("Nilai Tugas 2: "))
nilai3 = float(input("Nilai UTS: "))
nilai4 = float(input("Nilai Quiz: "))
nilai5 = float(input("Nilai UAS: "))

# Menghitung rata-rata atau total nilai mahasiswa (contoh: rata-rata dari 5 variabel)
nilai_mahasiswa = (nilai1 + nilai2 + nilai3 + nilai4 + nilai5) / 5

# Menentukan status kelulusan berdasarkan perbandingan logika
apakah_lulus = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan

# Menampilkan hasil
print(f"\n--- HASIL EVALUASI ---")
print(f"Nilai Mahasiswa (Rata-rata): {nilai_mahasiswa}")
print(f"Standar Kelulusan: {standar_kelulusan}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remedial? {perlu_remedial}")
```

```python
# 1. Mengubah variabel belanja agar menerima input melalui keyboard
# Menggunakan int() atau float() karena input() mengembalikan string
belanja = float(input("Masukkan total belanja: "))

# 2. Menyesuaikan diskon menjadi 35% (0.35) jika belanja di atas 1.000.000
# Menggunakan operator ternary / boolean indexing seperti pada contoh slide
diskon = (0, 0.35 * belanja) [belanja > 1000000]

# 3. Menghitung total harga akhir setelah diskon
total_harga = belanja - diskon

# Menampilkan hasil
print(f"Total Belanja: Rp {belanja}")
print(f"Diskon yang didapat: Rp {diskon}")
print(f"Total harga setelah diskon = Rp {total_harga}")
```
