```python

# ATS PRAKTIKUM RPE311 DASAR PEMROGRAMAN
# Nama         : Rian Saputra Nur
# NIM          : 41
# Tanggal      : 13 November 2026
# Mata Kuliah  : RPE311 Dasar Pemrograman
# Paket        : Genap

# SOAL 2
# Menghitung energi harian dan capacity factor PLTD

# Menyimpan data ke dalam variabel
daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400

# Menghitung energi harian
energi = daya_kw * jam

# Menghitung capacity factor
capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100

# Menampilkan hasil
print(f"Energi harian = {energi} kWh")
print(f"Capacity factor = {capacity_factor:.2f}%")
```

```python
# SOAL 6
# Menentukan status operasi boiler berdasarkan suhu dan tekanan

# Memasukkan data suhu dan tekanan
suhu = float(input("Masukkan suhu uap (C): "))
tekanan = float(input("Masukkan tekanan (bar): "))

# Menentukan status boiler
if suhu >= 550 or tekanan >= 14:
    print("STATUS: TRIP")

elif suhu > 520 and tekanan > 12:
    print("STATUS: WARNING")

else:
    print("STATUS: NORMAL")
```

```python
# SOAL 8
# Simulasi pengosongan tangki bahan bakar

# Volume awal tangki
volume = 100

# Perulangan selama tangki masih berisi bahan bakar
while volume > 0:

    # Mengurangi volume sebanyak 20 liter
    volume = volume - 20

    # Menampilkan volume saat ini
    print(f"Volume: {volume} liter")

    # Memeriksa apakah volume mencapai 20 liter
    if volume == 20:
        print("WARNING: Hampir kosong!")

# Menampilkan pesan setelah tangki kosong
print("TANGKI KOSONG - Pengosongan berhenti.")
```

```python
# SOAL 10
# Memeriksa suhu bearing dengan sensor rusak

# Data suhu bearing dalam bentuk teks
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

# Menghitung jumlah data valid
jumlah_valid = 0

# Memeriksa setiap data menggunakan perulangan for
for nomor, nilai in enumerate(data, start=1):

    try:
        # Mengubah data teks menjadi float
        suhu = float(nilai)

        # Menambah jumlah data valid
        jumlah_valid += 1

        # Memeriksa apakah suhu melebihi 70 derajat
        if suhu > 70.0:
            print(f"Warning: Titik {nomor} suhu bearing tinggi!")

    except ValueError:
        # Menampilkan pesan jika data tidak valid
        print(f"Titik {nomor}: data tidak valid - pemeriksaan dihentikan")

        # Menghentikan perulangan
        break

# Menampilkan jumlah data valid
print(f"Jumlah data valid: {jumlah_valid}")
```
